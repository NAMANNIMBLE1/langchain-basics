```python
def get_data():
    data = [
        {"input": "What is your name?", "output": "I am an AI assistant."},
        {"input": "How are you?", "output": "I'm doing well, thank you!"},
        {"input": "What is the capital of France?", "output": "Paris."},
        {"input": "What is 2 + 2?", "output": "4."},
        {"input": "Who is the president of the USA?", "output": "As of 2025, Joe Biden."},
        {"input": "What is the weather like today?", "output": "I can't check real-time weather, but it depends on your location."},
        {"input": "Tell me a joke.", "output": "Why did the scarecrow win an award? Because he was outstanding in his field!"},
        {"input": "What is the largest planet?", "output": "Jupiter."},
        {"input": "How many continents are there?", "output": "Seven."},
        {"input": "What is the boiling point of water?", "output": "100°C or 212°F."},
        {"input": "What language do you speak?", "output": "I can understand and respond in English."},
        {"input": "Can you help me with math?", "output": "Yes, I can help with basic to advanced math."},
        {"input": "What is AI?", "output": "AI stands for Artificial Intelligence."},
        {"input": "What is the color of the sky?", "output": "Usually blue during the day."},
        {"input": "What time is it?", "output": "I don't have real-time access to current time."},
        {"input": "What is Python?", "output": "Python is a programming language."},
        {"input": "How do I make tea?", "output": "Boil water, add tea leaves or a tea bag, steep, then serve."},
        {"input": "What is 10 divided by 2?", "output": "5."},
        {"input": "Where is the Great Wall of China?", "output": "In China."},
        {"input": "How many days are in a week?", "output": "Seven."}
    ]
    return data
```

```python
import lamini
from lamini import Lamini
import os
from dotenv import load_dotenv
load_dotenv()
lamini.api_key = os.getenv("LAMINI_API_KEY")
```

```python
llm = Lamini(model_name = "meta-llama/Meta-Llama-3-8B-Instruct")
```

```python
data = get_data()

""" common hyper parameters for all models 
    learning rate (string)
    early stopping (bool)
    max_steps (int)
    optimizer (string) optim 
"""

llm.tune(
    data_or_dataset_id=data,
    finetune_args={
        "learning_rate": 1e-5,
        "early_stopping": True,
        # "max_steps": 100,
        "optimizer": "adamw"
    }
)
```

```python
response = llm.generate("What is the capital of France?")
print("Response:", response)
```
